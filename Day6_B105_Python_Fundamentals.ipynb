```python
students=['chandu','poo','laddu','nandu','ashu','sukku']
print(students)
```

```python
msg=f'keep ur good work {students[0]}'
print(msg)
```

```python
for x in students:
    print(f'keep ur good work {x.title()}')
```

```python
for x in students:
    print(f'keep ur good work {x.title()}')
    print(f'do ur task today {x.title()}')
```

```python
for x in students:
    print(f'keep ur good work {x.title()}')
    print(f'do ur task today {x.title()}\n' )
```

```python
for x in students:
    print(f'keep ur good work {x.title()}')
    print(f'do ur task today {x.title()}\n')
print("thank you Students")
```
